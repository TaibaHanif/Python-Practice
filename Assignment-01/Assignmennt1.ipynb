## Task 1: Student Marks and Grades

```python
students = {
    "Ali": 85,
    "Sara": 72,
    "Ahmed": 35,
    "Ayesha": 91,
    "Usman": 55
}

def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"


passed_students = 0

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(name, marks, grade)

    if grade != "Fail":
        passed_students += 1

print("Number of passed students:", passed_students)
```

## Task 2: Shopping Bill Calculator

```python
products = {
    "Milk": 250,
    "Bread": 150,
    "Eggs": 400,
    "Rice": 1200
}

def calculate_discount(total):
    if total >= 5000:
        return total * 0.10
    else:
        return 0

print("Available Products:")

for name, price in products.items():
    print(name, "- Rs", price)

total_bill = 0

while True:
    item = input("Enter product name or 'done': ").title()

    if item == "Done":
        break

    if item in products:
        quantity = int(input("Enter quantity: "))
        total_bill += products[item] * quantity
    else:
        print("Product not found.")

discount = calculate_discount(total_bill)
final_bill = total_bill - discount

print("Original Bill: Rs", total_bill)
print("Discount: Rs", discount)
print("Final Bill: Rs", final_bill)
```

## Task 3: ATM Withdrawal System

```python
account_name = "Ali"
correct_pin = 1234
balance = 10000

attempts = 0
authenticated = False

while attempts < 3:
    entered_pin = int(input("Enter your PIN: "))

    if entered_pin == correct_pin:
        authenticated = True
        break
    else:
        attempts += 1
        print("Incorrect PIN")

if authenticated:
    amount = int(input("Enter withdrawal amount: "))

    def withdraw(balance, amount):
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal successful")
            return balance
        else:
            print("Insufficient balance")
            return balance

    balance = withdraw(balance, amount)
    print("Remaining balance: Rs", balance)

else:
    print("Maximum attempts reached. Account locked.")
```

## Task 4: Number Guessing Game

```python
secret_number = 7
max_attempts = 5

player_name = input("Enter your name: ")

score = 0
won = False

def check_guess(guess):
    if guess == secret_number:
        return "Correct"
    elif guess < secret_number:
        return "Too low"
    else:
        return "Too high"

for attempt in range(max_attempts):
    guess = int(input("Guess the number: "))
    result = check_guess(guess)

    print(result)

    if result == "Correct":
        score = max_attempts - attempt
        won = True
        break

player_result = {
    "name": player_name,
    "score": score
}

if won:
    print("Congratulations! You won.")
else:
    print("You used all attempts.")
    print("The correct number was:", secret_number)

print("Player result:", player_result)
```

## Task 5: Restaurant Ordering System

```python
menu = {
    "Biryani": 350,
    "Karahi": 1200,
    "Naan": 50,
    "Tea": 100
}

ordered_items = {}
total_bill = 0

print("Restaurant Menu:")

for item, price in menu.items():
    print(item, "- Rs", price)

while True:
    item = input("Enter item name or 'done': ").title()

    if item == "Done":
        break

    if item in menu:
        quantity = int(input("Enter quantity: "))

        if quantity > 0:
            ordered_items[item] = ordered_items.get(item, 0) + quantity

            item_total = menu[item] * quantity
            total_bill += item_total

            print("Item total: Rs", item_total)
        else:
            print("Quantity must be greater than zero.")
    else:
        print("Item is not available.")

print("Your order:", ordered_items)
print("Total bill: Rs", total_bill)
```

## Task 6: Employee Attendance and Salary

```python
employees = {
    "Ali": 26,
    "Sara": 23,
    "Ahmed": 18,
    "Ayesha": 30
}

monthly_salary = 30000
full_salary_count = 0

def calculate_salary(days):
    if days > 26 or days < 0:
        return "Invalid attendance"
    elif days == 26:
        return monthly_salary
    elif days >= 20:
        return monthly_salary * 0.80
    else:
        return monthly_salary * 0.50

for name, days in employees.items():
    salary = calculate_salary(days)

    print(name, "- Attendance:", days, "days - Salary:", salary)

    if salary == monthly_salary:
        full_salary_count += 1

print("Employees receiving full salary:", full_salary_count)
```
